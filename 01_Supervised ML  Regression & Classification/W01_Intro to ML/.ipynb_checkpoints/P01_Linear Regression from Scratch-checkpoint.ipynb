## r2_socre: coefficient of determination
- Is used to evaluate the performance of a linear regression model.
- It is the amount of the variation in the output dependent attribute that is predicted from the input independent variables
- Computed to asseess how well a regression model fits the observed data.
- It measures the proportion of the variance in the dependent variabel that is predictable from the independt variables.

## Using Core Python

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('canada_per_capita_income.csv')
df.head()
```

```python
X = df.drop('per capita income (US$)', axis=1)
y = df['per capita income (US$)']
```

```python
X = X.to_numpy()
y = y.to_numpy()
```

```python
X.reshape(-1)
```

```python
y
```

```python
X.shape
```

```python
y.shape
```

```python
class LinearRegression:
    def __init__(self, learning_rate=0.001, epochs=100):
        self.learning_rate = learning_rate
        self.epochs = epochs
        self.w = 0
        self.b = 0

    def fit(self, X, y):

        X = X.reshape(-1) # (n_samples, 1) -> (1, n_samples)
        
        # Scale X 
        self.X_mean = X.mean()
        self.X_std = X.std()
        X = (X-self.X_mean) / self.X_std
        
        self.num_samples = len(X)
        
        for epoch in range(self.epochs):
            dw = 0
            db = 0

            # Gradient computation
            for i in range(self.num_samples):
                f_wb = self.w * X[i] + self.b  # wx(i)+b
                dw += (f_wb - y[i]) * X[i]
                db += (f_wb - y[i])
            dw = dw/self.num_samples
            db = db/self.num_samples

            # Parameetrs update
            self.w -= self.learning_rate*dw
            self.b -= self.learning_rate*db

            # compute loss
            loss = 0
            for i in range(self.num_samples):
                # y = wX + b
                y_hat = self.w * X[i] + self.b
                loss += (y_hat - y[i])**2
            loss = loss / (2*self.num_samples)

            if epoch % 100 == 0:
                print(f'Epoch {epoch}; loss: {  loss:.4f}')
        
    def predict(self, X):
        X = np.array(X).flatten()

        # Apply same scaling
        X = (X - self.X_mean) / self.X_std

        return self.w * X + self.b
```

```python
learning_rate = 0.01
epochs = 100000
linear_model = LinearRegression(learning_rate, epochs)
linear_model.fit(X, y)
```

```python
y_pred = linear_model.predict(X)
```

```python
from sklearn.metrics import r2_score

r2_score(y, y_pred)
```

```python
linear_model.w, linear_model.b
```

```python
plt.xlabel('Year')
plt.ylabel('per capita income (US$)')
plt.scatter(X, y, marker='+')
plt.plot(X, y_pred)
```
