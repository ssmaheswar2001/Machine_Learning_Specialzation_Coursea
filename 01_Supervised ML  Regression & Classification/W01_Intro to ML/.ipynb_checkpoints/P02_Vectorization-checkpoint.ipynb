```python
# Dot Product

def my_dot(a, b):
    m = a.shape[0]

    dot = 0
    for i in range(m):
        dot += a[i]*b[i]
    return dot

a = np.array([1, 2, 3, 4])
b = np.array([-1, 4, 3, 2])
np.dot(a, b), my_dot(a, b)
```
