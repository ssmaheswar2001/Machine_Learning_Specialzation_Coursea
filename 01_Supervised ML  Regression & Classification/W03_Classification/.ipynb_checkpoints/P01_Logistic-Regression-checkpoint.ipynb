## Logistic Regression using scikit-learn

```python
import numpy as np
from sklearn.linear_model import LogisticRegression
```

```python
X = np.array([[0.5, 1.5], [1,1], [1.5, 0.5], [3, 0.5], [2, 2], [1, 2.5]])
y = np.array([0, 0, 0, 1, 1, 1])

X.shape, y.shape
```

```python
model = LogisticRegression()
model.fit(X, y)
```

```python
y_pred = model.predict(X)
y_pred
```

```python
model.score(X, y_pred)
```

```python
from sklearn.metrics import accuracy_score

accuracy_score(y, y_pred)
```

## Logistic Regression from Scratch

```python
import numpy as np
from matplotlib import pyplot as plt
```

```python
X = np.array([[0.5, 1.5], [1,1], [1.5, 0.5], [3, 0.5], [2, 2], [1, 2.5]])
y = np.array([0, 0, 0, 1, 1, 1])

X.shape, y.shape
```

```python
class LogisticRegression:
    def __init__(self, learning_rate=0.01, epochs=100):
        self.learning_rate = learning_rate
        self.epochs = epochs

    def sigmoid(self, z):
        return 1/(1+np.exp(- z))
    
    def compute_gradient(self, X, y):
        dw = np.zeros(self.n)
        db = 0.0

        for i in range(self.m):
            z = np.dot(X[i], self.w) + self.b
            f_x = self.sigmoid(z)
            err = f_x - y[i]
            for j in range(self.n):
                dw[j] += err*X[i, j]
            db += err
            
        dw = dw/self.m
        db = db/self.m

        return dw, db

    def compute_cost(self, X, y):
        cost = 0
        for i in range(self.m):
            z = np.dot(X[i], self.w) + self.b
            f_x = self.sigmoid(z)
            cost += -y[i]*np.log(f_x) - (1-y[i])*np.log(1-f_x)
        cost = cost/self.m
        return cost
    
    def fit(self, X, y):
        self.m, self.n =  X.shape

        self.w = np.zeros(self.n)
        self.b = 0

        for epoch in range(self.epochs):
            # Compute grdient
            dw, db = self.compute_gradient(X, y)
            
            # update the parameters using w, b, alpa and gradient
            self.w = self.w - self.learning_rate*dw
            self.b = self.b - self.learning_rate*db

            if epoch%100 == 0:
                print(f'Epoch {epoch}; cost: {self.compute_cost(X, y)}')

    def predict(self, X):
        z = np.dot(X, self.w) + self.b
        y_pred = self.sigmoid(z)
        return (y_pred >= 0.5)
```

```python
model = LogisticRegression(learning_rate=0.001, epochs=1000)
model.fit(X, y)
```
